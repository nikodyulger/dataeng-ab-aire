# Análisis Meteorológico Albacete

```python
import duckdb
import pandas as pd
import plotly.express as px

from utils import plot_boxplot_with_selector
```

```python
# Conectar a DuckDB y configurar conexión a MinIO
con = duckdb.connect()
con.execute("INSTALL httpfs;")
con.execute("LOAD httpfs;")
con.execute("SET s3_url_style='path';")
con.execute("SET s3_endpoint='localhost:9000';")
con.execute("SET s3_access_key_id='minioadmin';")
con.execute("SET s3_secret_access_key='minioadmin';")
con.execute("SET s3_use_ssl=false;")

# Leer todos los datos de MinIO para los años 2024-2026 y ambos tipos de parámetro
# tipo_parametros puede ser METEO o CONTAMINANTE
query = """
SELECT *
FROM read_parquet('s3://oro/tipo_parametros=METEO/year=*/month=*/consolidated_data.parquet')
"""
df = con.execute(query).fetchdf()

duckdb.sql("COPY (select * from df) to 'meteo_ab.csv'")

con.close()

df.head()
```

```python
print("Columnas del DataFrame:")
print(df.columns.tolist())

print("\nInformación del DataFrame:")
df.info()
```

```python
cols_interes = ['temp_c', 'ruido_db', 'vel_viento', 'pres']

for col in cols_interes:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))]
    print(f"Anomalías detectadas en {col}: {len(outliers)}")
```

Parece que hay ciertos datos anómalos para cada una de las variables siendo la presion y la velocidad del viento con el mayor numero de *outliers*

# Visualizaciones

Veamos mejor la distribución de las variables principales, así podremos identificar mejor esos *outliers*

```python
vars_analisis = ['temp_c', 'ruido_db', 'vel_viento', 'pres']

unidades = {
    'temp_c': '°C',
    'ruido_db': 'dB',
    'vel_viento': 'm/s',
    'pres': 'hPa'
}
fig = plot_boxplot_with_selector(df, vars_analisis, unidades)
fig.show()
```

La **temperatura** y la **presion** son las variables que más semejanza tienen entre las diferentes estaciones meteorológicas.
Sin embargo, no pasa lo mismo con la velocidad del **viento** y el **ruido** donde a pesar de que los valores medios se asemejen, hay estaciones con mas *outliers* y disparidad que otras:
- La estación en la Avenida Isabel La Católica (Isleta) o comunmente denominada como *Veleta* es la que mas variación de ruido presenta. ¿será cosa del sensor o por algun evento esporádico?
- Las mayores rachas de viento se han registrado en la estación de la Carretera de Madrid Esquina Cronista o comunmente llamada *Rotonda de la Tamos* que es un lugar que se caracteriza porque es la entrada a la ciudad. No hay tanto edificios o árboles de parques que suavicen el viento. Cabe añadir en este punto que los *outliers* guardan en todas las estaciones tienen el mismo comportamiento, eso puede indicar los días de mas viento.

Empecemos por el analisis univariado, por excelencia cogeremos la *temperatura* como variable de referencia aunque tambien se puede con el resto, tan solo hay que cambiar el eje de las *Y*

```python
fig = px.line(
    df.sort_values(['station_slug', 'fecha']),
    x='fecha',
    y='temp_c',
    color='station_slug',
    title='Evolución de la temperatura por estación',
    labels={'temp_c': 'Temperatura (°C)', 'fecha': 'Fecha', 'station_slug': 'Estación'}
)
fig.update_layout(legend_title_text='Estación')
fig.show()
```

Se puede apreciar la estacionalidad de la temperatura, es decir, **baja** en los meses de invierno y **sube** en los meses de verano a lo largo de los años. Veamos la gráfica quitando algo del ruido dado que los registros son horarios y se comprime demasiada información

```python
df_daily = (
    df.set_index('fecha')
      .groupby('station_slug')['temp_c']
      .resample('D')
      .mean()
      .reset_index()
)

fig = px.line(
    df_daily.sort_values(['station_slug', 'fecha']),
    x='fecha',
    y='temp_c',
    color='station_slug',
    title='Evolución diaria promedio de la temperatura por estación',
    labels={'temp_c': 'Temperatura media (°C)', 'fecha': 'Fecha', 'station_slug': 'Estación'}
)
fig.update_layout(legend_title_text='Estación')
fig.show()
```

¡Mucho mejor! Se puede apreciar con más detalle las ligeras diferencias en las mediciones entre las estaciones

En cambio si agrupamos por hora, obtenemos lo siguiente:

```python
df["hour"] = df["fecha"].dt.hour

temp_diurna = df.groupby(["station_slug", "hour"])["temp_c"].mean().reset_index()

fig = px.line(
    temp_diurna,
    x="hour",
    y="temp_c",
    color="station_slug",
    title="Perfil medio de temperatura a lo largo del día"
)
fig.show()
```

La evolución de la temperatura a lo largo del día es bastante intuitiva: "De noche hace más frío y conforme el sol sube mas y mas en el cielo, hace mas calorcito"

```python
# Limpiamos los datos 
df_locos = df.dropna(subset=['temp_c_daily_std'])
df_locos = df_locos[df_locos['temp_c_daily_std'] > 0]

# Agrupamos 
df_locos_daily = df_locos.groupby([df_locos['fecha'].dt.date, 'station_slug'])['temp_c_daily_std'].max().reset_index()

fig = px.scatter(df_locos_daily, x='fecha', y='temp_c_daily_std',
                 size='temp_c_daily_std',
                 color='station_slug',
                 hover_name='station_slug',
                 title='Días de alta disparidad térmica (Inestabilidad)',
                 labels={'temp_c_daily_std': 'Grados de Variación (Std Dev)'})

fig.show()
```

Usando una de las características precalculadas podemos ver la anterior gráfica como evoluciona a lo largo del año. Cuanto más grande es el punto, significa que la anterior gráfica tiene un mayor rango de valores (en el eje Y)
Hablando en cristiano, podemos identificar los días en los que la temperatura ha tenido mayor variación. Justo esos días en los que hacen falta capas de ropa diferentes porque no sabes si hace calor o frío

Se aprecia la estacionalidad, los días más anomalos son los de primavera, verano y otoño. En invierno ya sabemos que hace frío, en verano por las noches refresca y por el día te mueres de calor

La misma gráfica de antes, también se puede poner de otra forma para ver mejor la amplitud del rango. 
`15` o `20` grados de diferencia me parece mucho

```python
temp_range = df.groupby(
    [df["fecha"].dt.date, "station_slug"]
)["temp_c"].agg(lambda x: x.max() - x.min()).reset_index(name="range")

fig = px.line(
    temp_range,
    x="fecha",
    y="range",
    color="station_slug",
    title="Rango térmico diario"
)
fig.show()
```

```python
df_diario = df.groupby(df['fecha'].dt.date)[['temp_c', 'wind_chill']].mean().reset_index()

# 2. Creamos la gráfica
fig = px.line(df_diario, x='fecha', y=['temp_c', 'wind_chill'],
              title='Evolución de Temperatura vs. Sensación Térmica (2024-2026)',
              labels={'value': 'Grados Celsius (°C)', 'variable': 'Medición', 'fecha': 'Fecha'},
              color_discrete_map={'temp_c': '#EF553B', 'wind_chill': '#636EFA'})

# Mejoramos la interactividad
fig.update_layout(hovermode="x unified", legend_title_text='Variable')
fig.show()
```

Algo que costó tiempo encontrar una fórmula que se entiendese fue la de la **sensación térmica** la cual solo se calcula para temperaturas por debajo de `10ºC` de ahi este marcada en azul los meses de invierno y tal como era de esperar está por debajo de la temperatura "real" y eso se debe al viento

Una de las preguntas que quiero contestar es si Albacete es realmente una ciudad ventosa

--> La Escala Beaufort se utiliza principalmente para medir la intensidad del viento (normalmente en mar, pero lo aplicaremos también en secano)

```python
beaufort_labels = {0: 'calma', 1: 'ventolina', 2: 'flojito', 3: 'flojo', 4: 'bonancible'}
df['beaufort_label'] = df['beaufort'].map(beaufort_labels).fillna(df['beaufort'].astype(str))
fig = px.histogram(
    df,
    x='beaufort',
    color='beaufort_label',
    title='Distribución de la escala Beaufort',
    labels={'beaufort': 'Beaufort', 'beaufort_label': 'Descripción'},
    category_orders={'beaufort': [0, 1, 2, 3, 4]},
    text_auto=True
)
fig.update_xaxes(tickmode='array', tickvals=list(beaufort_labels.keys()), ticktext=list(beaufort_labels.values()))
fig.show()
```

La mayor parte de los días hay algo de vientelina y luego hay ciertos días que sopla "flojo"

```python
df_daily = (
    df.set_index('fecha')
      .groupby('station_slug')['vel_viento']
      .resample('D')
      .mean()
      .reset_index()
)

fig = px.line(
    df_daily.sort_values(['station_slug', 'fecha']),
    x='fecha',
    y='vel_viento',
    color='station_slug',
    title='Evolución diaria promedio de la velocidad del viento por estación',
    labels={'vel_viento': 'Velocidad media del viento (km/h)', 'fecha': 'Fecha', 'station_slug': 'Estación'}
)
fig.update_layout(legend_title_text='Estación')
fig.show()
```

- Confirmamos lo de que aqui sopla viento todo el año
- Parece que la estación del Paseo de Cuba Fabrica de HArinas es la que menores valores registra y tal como vimos con los *outliers* la estación en la Rotonda de la Tamos (Carretera de Madrid) es la que mayores rachas pilla

Veamos si con el ruido ocurre la misma discrepancia entre las estaciones meteorológicas

```python
day_labels = {0: 'Lunes', 1: 'Martes', 2: 'Miércoles', 3: 'Jueves', 4: 'Viernes', 5: 'Sábado', 6: 'Domingo'}

df['weekday'] = df['fecha'].dt.weekday
ruido_weekday = (
    df.groupby(['weekday', 'station_slug'], sort=False)['ruido_db']
        .mean()
        .reset_index()
)
fig = px.bar(
    ruido_weekday,
    x='weekday',
    y='ruido_db',
    color='station_slug',
    barmode='group',
    title='Ruido medio por día de la semana y estación',
    labels={'weekday': 'Día de la semana', 'ruido_db': 'Ruido medio (dB)', 'station_slug': 'Estación'}
)
fig.update_xaxes(tickmode='array', tickvals=list(day_labels.keys()), ticktext=list(day_labels.values()))
fig.show()
```

A vista de semana, todos los días parecen iguales. Eso indica calma, creía que quizás los días entre semana seríamos algo mas ruidosos por los coches yendo a trabajar, pero no señor.

Eso sí, la estación da la Carretera de Madrid es una de las más ruidosas y tiene todo el sentido del mundo, es donde más tráfico de coches hay.
```python
df['hora'] = df['fecha'].dt.hour

df_ruido = df.groupby(['station_slug', 'hora'])['ruido_db'].mean().reset_index()

fig = px.line(df_ruido, x='hora', y='ruido_db', color='station_slug',
              title='Nivel de Ruido Promedio por Hora del Día',
              labels={'ruido_db': 'Decibelios (dB)', 'hora': 'Hora del día'})
fig.show()
```

Es misma diferencia, trasladada a las horas del día también se puede visualizar. Y al igual que la temperatura se ve la misma evolución a lo largo del día y todo es por la misma razón: el sol. Por la noche, la gente duerme y por el día, la gente se activa y hace ruido

Veamos una última gráfica que nos puede valer a abrir el melón de qué más se podría hacer con estos datos: modelo de machine learning

--> El punto de rocío es la temperatura a la cual se debe enfriar el aire para que el vapor de agua se condense en rocío o escarcha.

Por la propia definición y de las anteriores fases del proyecto, sabemos que la variable `punto_rocio` la calculamos a partir de la `temp_c` y la `humedad` por tanto lo que esperamos ver con este gráfico es justo esa relación entre las variables. Esa misma relación es lo que se determina como regresión y podríamos predecirla.

```python
fig = px.scatter(df[df['station_slug']== "plaza_isabel_ii"], x='temp_c', y='humedad', color='punto_rocio',
                 title='Relación entre Temperatura, Humedad y Punto de Rocío',
                 color_continuous_scale='Viridis',
                 labels={'temp_c': 'Temperatura (°C)', 'humedad': 'Humedad (%)'})
fig.show()
```

Se aprecia a simple vista que: 
- A mayor temperatura, a partir de cierto nivel de humedad se llega al punto de rocío con poco
- Si hay demasiada humedad, no hace falta mucha temperatura para llegar a que se condense
